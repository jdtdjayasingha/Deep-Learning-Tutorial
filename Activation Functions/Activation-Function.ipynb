Sigmoid Function

```python
import numpy as np
```

```python
def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))
```

```python
inputs = [13, 3, -12]
```

```python
for x in inputs:
    print('{} --> {}'.format(x,sigmoid(x)))
```

Tanh Function

```python
import numpy as np
```

```python
def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))
```

```python
inputs = [13, 3, -12]
```

```python
for m in inputs:
    print('{} --> {}'.format(m,tanh(m)))
```
