Sigmoid Function

```python
import numpy as np
```

```python
def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))
```

```python
inputs = [13, 3, -12]
```

```python
for x in inputs:
    print('{} --> {}'.format(x,sigmoid(x)))
```

Tanh Function

```python
import numpy as np
```

```python
def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))
```

```python
inputs = [13, 3, -12]
```

```python
for m in inputs:
    print('{} --> {}'.format(m,tanh(m)))
```

ReLU Function

```python
import numpy as np
```

```python
def relu(x):
    return max(0.0,x)
```

```python
inputs = [13, 3, -12]
```

```python
for x in inputs:
    print('{} --> {}'.format(x,relu(x)))
```

Leaky ReLU Function

```python
import numpy as np
```

```python
def leakyRelu(x):
    return max(0.1 * x, x)
```

```python
inputs = [13, 3, -12]
```

```python
for x in inputs:
    print('{} --> {}'.format(x, leakyRelu(x)))
```

Softmax Funcion

```python
import numpy as np
```

```python
def softmax(x):
    return (np.exp(x) / np.exp(x).sum())
```

```python
inputs = [1, 3, -2]
```

```python
print (softmax(inputs))
```
